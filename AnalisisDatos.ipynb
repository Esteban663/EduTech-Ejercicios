```python
import pandas as pd
```

```python
#Importación Limpia del Archivo:

df_plataforma = pd.read_csv("estudiantes_cursos.csv")
df_plataforma
```

```python
# Extracción y Proyección de Atributos
# Extrae la columna Puntaje_Evaluacion como una Serie de Pandas

pd.Series(df_plataforma['Puntaje_Evaluacion'])
```

```python
# Extracción y Proyección de Atributos
# Extrae un nuevo subconjunto bidimensional con las columnas Curso, Modalidad y Lecciones_Completadas.

df_subconjunto = df_plataforma[['Curso', 'Modalidad', 'Lecciones_Completadas']]
df_subconjunto
```

```python
# Obtén todos los atributos del estudiante ubicado específicamente en el índice 120 utilizando el indexador .loc[].


estudiante_120 = df_plataforma.loc[[120]]
estudiante_120
```

```python
#  Identifica y elimina todas las filas duplicadas dentro del DataFrame utilizando el método .drop_duplicates().
df_sin_duplicados = df_plataforma.drop_duplicates()
df_sin_duplicados
```

```python
#* Realiza un filtro booleano que devuelva únicamente aquellos estudiantes que hayan completado 
# más de 40 lecciones (Lecciones_Completadas > 40).

def Lecciones_Completadas_mayores_a_40(df):
    return df[df['Lecciones_Completadas'] > 40]

df_filtrado = Lecciones_Completadas_mayores_a_40(df_sin_duplicados)
df_filtrado
```

```python
#* Filtrado por Coincidencia Múltiple (isin): 
# Emplea el método .isin() para seleccionar únicamente los registros 
# de estudiantes inscritos en los cursos de Python para Datos y Arquitectura Cloud.

df_cursos_seleccionados = df_filtrado[df_filtrado["Curso"].isin(["Python para Datos", "Arquitectura Cloud"])]
df_cursos_seleccionados
```

```python
# Filtrado por Rango de Desempeño (between): Utilizando el método .between(), extrae todos los alumnos cuyo
#  Puntaje_Evaluacion se encuentre entre 70.0 y 85.0 inclusive.

rango_desempeño = df_filtrado[df_filtrado['Puntaje_Evaluacion'].between(70.0, 85.0)]
rango_desempeño
```

```python
# Consultas Legibles con .query(): Aplica una consulta con .query() para filtrar aquellos registros
# cuya Modalidad sea igual a "Asincrónico" y sus Horas_Estudio superen las 25 horas.

df_consulta = df_filtrado.query('Modalidad == "Asincrónico" and Horas_Estudio > 25')
df_consulta
```

```python
# Filtrado Compuesto con Búsqueda de Texto: Combina condiciones lógicas con el operador & para localizar
#  a los estudiantes cuyo Nivel_Educativo sea "Profesional" y cuyo nombre de Curso inicie con la letra "B"
#  usando el accesor .str.startswith().

filtro_compuesto = (df_filtrado["Nivel_Educativo"] == "Profesional") & (df_filtrado["Curso"].str.startswith("B"))
df_filtrado[filtro_compuesto]
```

```python
import numpy as np
```

```python
# Imputación de Datos Faltantes con Estrategias Mixtas: 
# Reemplaza los valores nulos (NaN) en la columna Horas_Estudio por el promedio aritmético de dicha serie. 
# En la columna Puntaje_Evaluacion, reemplaza los valores ausentes asignando una nota fija de 0.0 mediante el método .fillna().

df_rellenado = df_filtrado.fillna({
    "Horas_Estudio": df_filtrado["Horas_Estudio"].mean(),
    "Puntaje_Evaluacion": 0.0
})

df_rellenado
```

```python
# Transformación Vectorizada con .apply(): 
# Define una función en Python que reciba la columna Puntaje_Evaluacion y devuelva el texto "Aprobado" 
# si la nota es mayor o igual a 70.0, o "Reprobado" en caso contrario. Utiliza .apply() 
# para crear una nueva columna denominada Estado_Materia.

def Aprobacion_Puntaje(Puntaje):
    if (Puntaje >= 70.0):
        return "APROBADO"
    else:
        return "REPROBADO"


df_rellenado["Estado_Materia"] = df_rellenado["Puntaje_Evaluacion"].apply(Aprobacion_Puntaje)
df_rellenado
```

```python
# Estructuración y Renombrado: 
# Renombra la columna Modalidad a Formato_Clase y Certificado_Generado a Graduado usando .rename(). 
# Luego, reordena las columnas para que Estado_Materia e ID_Estudiante sean las dos primeras.

Renombrar_columnas = df_rellenado.rename(columns={
    "Modalidad": "Formato_Clase",
    "Certificado_Generado": "Graduado"
})
Renombrar_columnas
```

```python
# Columnas Ordenadas: Reorganiza las columnas del DataFrame para que Estado_Materia e ID_Estudiante sean las dos primeras

Ordenar_columnas = Renombrar_columnas[["Estado_Materia", "ID_Estudiante", "Curso", "Formato_Clase", "Lecciones_Completadas", "Horas_Estudio", "Puntaje_Evaluacion", "Graduado"]]
Ordenar_columnas
```

```python
# Agrupación y Estadísticas Globales por Programa: 
# Agrupa el DataFrame por Curso y, empleando el método .agg(), calcula el promedio de Horas_Estudio y 
# la suma acumulada de Lecciones_Completadas para cada programa académico.

Agrupacion_estadisticas = Ordenar_columnas.groupby("Curso").agg({ "Horas_Estudio": "mean", "Lecciones_Completadas": "sum" })
Agrupacion_estadisticas
```

```python
# Cálculo de Dispersión con Función Personalizada: 
# Crea una función llamada rango que devuelva la amplitud de una serie (series.max() - series.min()). 
# Aplica esta función con .agg() para calcular la variación o rango del Puntaje_Evaluacion agrupado por Nivel_Educativo.

def rango(series):
    return series.max() - series.min()

agrupacion_datos = Renombrar_columnas.groupby("Nivel_Educativo").agg({
    "Puntaje_Evaluacion": rango
})

agrupacion_datos
```

```python
# Agrupamiento Multinivel con Métricas Diferenciadas: 
# Agrupa simultáneamente los datos por Curso y Formato_Clase (o Modalidad), 
# obteniendo para cada subgrupo la suma de Horas_Estudio y el promedio final del Puntaje_Evaluacion.

agrupamiento_multinivel = Renombrar_columnas.groupby(["Curso", "Formato_Clase"]).agg({
    "Horas_Estudio": "sum",
    "Puntaje_Evaluacion": "mean"
})
agrupamiento_multinivel
```
